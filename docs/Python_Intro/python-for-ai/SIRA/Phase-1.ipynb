Would you write

```python
pd.read_csv(...)
```

inside every file?

No.

Every project has one place responsible for loading data.

This follows the **Single Responsibility Principle**:

> A module should have one responsibility.

Today, we don't have a CSV yet. So we'll simulate one.

Create:

```text
src/

data_loader.py
```

```python
# data_loader.py

"""Data Loader module"""

def load_sample_reports():
    """
    Simulates loading incident reports.

    Later this function will read a CSV file.
    """

    reports = [
        "Pressure leak detected on Pipeline 7.",
        "Worker slipped near the drilling platform.",
        "Gas detector triggered an alarm."
    ]

    return reports

```

### Step 6: Creating our first `Class`

Now let's connect OOP.

Create:

```text
src/

incident.py
```

Class:

```python
# incident.py

class Incident:
    """
    Represents a single incident report.

    Every incident in our system will become an object of this class.
    """

    def __init__(self, incident_id, report_text):
        self.incident_id = incident_id
        self.report_text = report_text

    def display(self):
        print("-" * 50)
        print(f"Incident ID : {self.incident_id}")
        print(f"Report      : {self.report_text}")
```

Explaination:

Each incident report becomes an object.

Later:

```python
incident1

incident2

incident3
```

Instead of dictionaries everywhere.

---

Example:

```python
from src.data_loader import load_data

load_data()
```

Now inside `main.py` add:

```python
from src.data_loader import load_sample_reports
from src.incident import Incident


def main():
    print("=" * 60)
    print("SMART INCIDENT REPORT ANALYZER (SIRA)")
    print("=" * 60)

    reports = load_sample_reports()

    incidents = []

    for index, report in enumerate(reports, start=1):
        incident = Incident(index, report)
        incidents.append(incident)

    print(f"\nLoaded {len(incidents)} incident reports.\n")

    for incident in incidents:
        incident.display()


if __name__ == "__main__":
    main()
```

### Step 9: Build the first running application

Create `__init__.py` file.

Note: `__init__.py` file is not neccessary but its still often use beacause of its role.

Explanation:

### Why does it exist?

Suppose you have

```
src/

    incident.py

    data_loader.py
```

Python sees them as ordinary files.

But if you add

```
src/

    __init__.py
```

Python now recognizes `src` as a **package**.

Now you can write

```python
from src.incident import Incident
```

Without `__init__.py`, older versions of Python wouldn't recognize the folder as a package. Modern Python (3.3+) supports namespace packages, but **including `__init__.py` is still a widely adopted best practice** because it makes the package explicit, improves compatibility, and provides a place for package-level initialization if you need it later.

### Think of it like this

Imagine a library.

Books are placed inside shelves.

The shelf has a label:

```
Science
```

Without the label,

people don't know what the shelf represents.

`__init__.py` is that label.
