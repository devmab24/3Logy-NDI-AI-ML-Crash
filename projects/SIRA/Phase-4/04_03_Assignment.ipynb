### Industry insight

Introduce the idea of a **baseline model**:

1. Train a simple model (e.g., Logistic Regression).
2. Measure its performance.
3. Train stronger alternatives.
4. Compare objectively.
5. Choose the model that best balances accuracy, speed, interpretability, and maintenance.

This is how production ML teams work. They don't assume a complex model is automatically better.
