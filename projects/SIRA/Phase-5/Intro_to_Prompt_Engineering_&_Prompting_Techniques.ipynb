## Introduction to Prompt Engineering & Prompting Techniques

**Project:** Smart Incident Report Analyzer (SIRA)

2. Understand the anatomy of an effective prompt.
3. Distinguish zero-shot, one-shot, and few-shot prompting.
4. Use role, context, constraint, and structured-output prompting.

### 1. What is Generative AI?

**Generative Artificial Intelligence (Generative AI)** refers to AI systems that can generate new content from instructions or prompts.

Outputs can include:

- Text
- Code
- Images
- Audio
- Video
- Structured data

Large Language Models (LLMs) are Generative AI systems designed primarily to understand and generate language.

Common LLM tasks include:

- Summarization
- Question answering
- Classification
- Information extraction
- Translation
- Report generation
- Code generation

**Traditional ML vs. LLM-based analysis**

Current SIRA:

```text
Incident Report
      ↓
Text Preprocessing
      ↓
TF-IDF
      ↓
ML Classifier
      ↓
Incident Type
```

Future GenAI-enhanced SIRA:

```text
Incident Report
      ↓
Prompt + Context
      ↓
Foundation Model / LLM
      ↓
Incident Analysis
      ↓
Structured Result
```

The two approaches can work together rather than one necessarily replacing the other.

---

### 2. What is a Prompt?

A **prompt** is the input or instruction provided to a Generative AI model to guide its response. Simple example:

```text
    Summarize this incident report.
```

More specific example:

```text
    You are an HSE incident analyst.

    Analyze the following workplace incident report.

    Identify:
    1. Incident type
    2. Severity
    3. Likely cause
    4. Recommended immediate action

    Do not invent facts that are not present in the report.

    Incident:
    An employee slipped on a wet floor in the warehouse.
```

The second prompt gives clearer instructions about the task, context, and expected output.

---

### 3. What is Prompt Engineering?

**Prompt engineering** is the process of designing, testing, and improving prompts to obtain useful, reliable, and appropriately structured outputs from Generative AI models.

A practical workflow is:

```text
     Define the task
          ↓
     Write an initial prompt
          ↓
     Run the prompt
          ↓
     Inspect the output
          ↓
     Identify weaknesses
          ↓
     Improve the prompt
          ↓
     Test again
```

Prompt quality can affect:

- Task interpretation
- Relevance of the answer
- Output structure
- Consistency
- Whether the model follows constraints

---

### 4. Anatomy of a good Prompt

A useful framework is:

> **Role + Task + Context + Constraints + Output Format**

Not every prompt needs every component.

- **Role**

Tell the model what perspective or function it should perform.

```text
    You are an HSE incident analyst.
```

- **Task**

Clearly state what the model needs to do.

```text
    Classify the incident and identify the likely cause.
```

- **Context**

Provide relevant background.

```text
    The incident occurred in an industrial warehouse.
```

- **Constraints**

Define boundaries.

```text
    Do not invent information that is not contained in the report.
```

- **Output Format**

Specify how the answer should be returned.

```text
    Incident Type:
    Severity:
    Likely Cause:
    Recommended Action:
```

## Types of prompting

### 1. Zero-Shot Prompting

**Zero-shot prompting** means asking a model to perform a task **without providing examples**.

Example:

```text
    Classify this workplace incident into one of these categories:

    - Fire
    - Slip/Fall
    - Vehicle Accident
    - Chemical Spill
    - Equipment Failure

    Incident:
    An employee slipped on a wet floor while walking through the warehouse.

    Category:
```

The model must understand the task from the instruction alone.

**When is it useful?**

- The task is straightforward.
- The model already understands the task.
- You do not have suitable examples.
- You want a concise prompt.

```text
    Instruction + New Input → Model → Output
```

---

### 2. One-Shot Prompting

**One-shot prompting** provides **one example** before asking the model to perform the task.

Example:

```text
     You are a workplace incident classifier.

     Classify incidents into:
          Fire, Slip/Fall, Vehicle Accident, Chemical Spill, Equipment Failure.

     Example:

     Incident:
          A worker slipped on a wet warehouse floor.

     Category:
          Slip/Fall

     Now classify:

     Incident:
          A forklift collided with a storage rack.

     Category:
```

The example demonstrates the expected relationship between input and output.

**Comparison**

**Zero-shot**

```text
     Instruction → New incident → Answer
```

**One-shot**

```text
     Instruction
          +
     Example
          +
     New incident
          ↓
     Answer
```

---

### 3. Few-Shot Prompting

**Few-shot prompting** provides several examples before the new input.

Example:

```text
    You are a workplace incident classifier.

    Classify incidents into:
    Fire, Slip/Fall, Vehicle Accident, Chemical Spill, Equipment Failure.

    Example 1:
    Incident: A worker slipped on a wet floor.
    Category: Slip/Fall

    Example 2:
    Incident: A forklift struck a storage rack.
    Category: Vehicle Accident

    Example 3:
    Incident: A pipe released hazardous liquid onto the floor.
    Category: Chemical Spill

    Example 4:
    Incident: Smoke was observed from an electrical panel.
    Category: Fire

    Now classify:

    Incident:
    A worker fell after losing balance near a wet processing area.

    Category:
```

Examples can help communicate the task, labels, style, and expected input-output relationship.

**Note**

More examples are not automatically better. Examples should be:

- Relevant
- Clear
- Representative
- Correct
- Consistent

---

### Zero-Shot, One-Shot, and Few-Shot

| Technique | Examples | Main Purpose |
|---|---:|---|
| Zero-shot | 0 | Give instructions only |
| One-shot | 1 | Demonstrate the task once |
| Few-shot | Several | Demonstrate a pattern using multiple examples |

Think of the progression:

```text
    ZERO-SHOT
    "Do this task."

    ONE-SHOT
    "Do this task. Here is one example."

    FEW-SHOT
    "Do this task. Here are several examples."
```

The goal is not to use as many examples as possible. The goal is to use the **smallest useful set of high-quality examples** that communicates the desired behavior.

## 4. Role Prompting

**Role prompting** tells the model what role or perspective it should adopt.

Example:

```text
    You are an experienced HSE incident analyst working in an industrial facility.
```

Compare:

```text
    Analyze this incident.
```

with:

```text
    You are an HSE incident analyst.

    Analyze this workplace incident and identify the incident type,
    possible cause, severity, and recommended immediate action.
```

Possible SIRA roles include:

- HSE Incident Analyst
- Safety Officer
- Incident Report Reviewer
- Compliance Analyst
- Industrial Risk Analyst

A role should support the task rather than simply make the prompt longer.

---

### 5. Context Prompting

**Context** provides relevant background information.

Example:

```text
    Context:
    The company operates an industrial processing facility.
    The incident occurred during the morning shift.
    The affected area contains heavy machinery and chemical processing equipment.

    Incident:
    A worker reported a strong chemical smell near the processing line.
```

**Avoid irrelevant context**

Bad context includes unrelated facts such as company history or office details when the task is incident classification.

Good prompt engineering means providing **relevant context**, not simply more text.

---

### 6. Constraint Prompting

Constraints tell the model how to behave or what boundaries to follow.

Example:

```text
    Classify the incident using ONLY one of these categories:

    Fire
    Slip/Fall
    Vehicle Accident
    Chemical Spill
    Equipment Failure

    Do not create a new category.
    Do not invent facts.
    If the report does not contain enough information, return:
    "Insufficient Information".
```

Constraints can reduce unwanted behavior such as:

- Inventing categories
- Adding unsupported information
- Returning unnecessarily long responses
- Ignoring required formats

Constraints are especially important when model outputs will be consumed by software.

---

### 7. Structured Output Prompting

Instead of requesting a general response, specify the structure you need.

Example:

```text
  Analyze this incident.

  Return exactly these fields:

  Incident Type:
  Severity:
  Likely Cause:
  Recommended Action:
```

A machine-friendly format can be:

```text
  Return valid JSON with exactly these keys:

  {
    "incident_type": "",
    "severity": "",
    "likely_cause": "",
    "recommended_action": ""
  }
```

**Why it matters**

Structured output makes responses easier for applications to consume through:

- Python
- APIs
- Databases
- Streamlit
- Reports
- Other ML/AI components

---

### 8. Instruction Prompting

Instruction prompting focuses on giving the model a clear, direct task.

Weak:

```text
    Incident analysis.
```

Better:

```text
    Analyze the incident report and identify the most likely incident type.
```

Better still:

```text
    Analyze the incident report.

    1. Identify the most likely incident type.
    2. Identify the likely cause using only information in the report.
    3. Assign a severity level of Low, Medium, or High.
    4. Recommend one immediate safety action.
```

**Principle**

Do not make the model guess what the task is.

**State the task explicitly.**

---

### 9. Prompt Iteration

Prompt engineering is iterative.

Weak prompt:

```text
    Analyze this incident:
        A worker fell near a machine.
```

Potential problems:

- What does "analyze" mean?
- Which incident categories are allowed?
- Should severity be included?
- Should the model recommend an action?
- Should it use information outside the report?

Improved prompt:

```text
    You are an HSE incident analyst.

    Analyze the following workplace incident.

    Incident:
    A worker fell near a machine while moving materials.

    Determine:
    1. Incident type
    2. Severity: Low, Medium, or High
    3. Likely cause
    4. Recommended immediate safety action

    Rules:
    - Use only information provided in the incident.
    - Do not invent facts.
    - If information is insufficient, state this clearly.

    Return the result using the requested fields.
```

**Engineering mindset**

Instead of asking:

> "Why did the model give me a bad answer?"

Ask:

> "What information or instruction was missing from my prompt?"

### Comparison

| Technique                            | What to learn                        | SIRA example                                  |
| ------------------------------------ | ------------------------------------ | --------------------------------------------- |
| **Zero-shot prompting**              | Give an instruction without examples | Classify an unfamiliar incident               |
| **One-shot prompting**               | Give one example                     | Show one incident → category                  |
| **Few-shot prompting**               | Give several examples                | Provide 3–5 categorized incidents             |
| **Role prompting**                   | Define the model's role              | "You are an HSE incident analyst..."          |
| **Structured prompting**             | Specify expected output              | Return category, severity, reasoning          |
| **Chain-of-thought-style reasoning** | Ask model to reason before answering | Analyze evidence before classification        |
| **Constraint prompting**             | Restrict what the model can output   | Only select from approved incident categories |

### Connecting Prompt Engineering to SIRA

Current SIRA:

```text
      Incident Report
            ↓
      Preprocessing
            ↓
      TF-IDF
            ↓
      ML Classifier
            ↓
      Incident Classification
```

GenAI can add capabilities such as:

```text
      Incident Report
            ↓
      Prompt Template
            ↓
      Foundation Model
            ↓
      ┌───────────────┬──────────────┐
      ↓               ↓              ↓
      Summary       Explanation   Recommendations
```

Eventually, both approaches can be combined:

```text
                         SIRA
                          │
              ┌───────────┴───────────┐
              ↓                       ↓
       Traditional ML              GenAI
              │                       │
       Incident Type          Explanation/Summary
              │                       │
              └───────────┬───────────┘
                          ↓
                   Unified Analysis
```

**Prompt Engineering best practices**

1. Be specific: State exactly what the model should do.

2. Provide relevant context: Give the information the model actually needs.

3. Define constraints: Specify boundaries and allowed values.

4. Give examples when useful: Use one-shot or few-shot prompting when examples clarify the expected behavior.

5. Specify the output format: Especially when another application will consume the response.

6. Avoid ambiguity: Words such as "analyze" or "improve" may be too vague without further instructions.

7. Do not assume private context: Provide information required for the task.

8. Test multiple cases: A prompt that works for one incident may fail on another.

9. Use representative examples: Few-shot examples should resemble real inputs.

10. Treat prompts as application components: In production, prompts should be versioned, tested, reviewed, and maintained like other software assets.


---

### Important limitations

Prompt engineering does **not** guarantee that an LLM will always produce correct answers.

LLMs can:

- Make factual errors
- Misinterpret ambiguous instructions
- Produce inconsistent outputs
- Generate unsupported information
- Follow an incorrect example
- Fail on unusual inputs

Therefore:

> **A good prompt improves model behavior; it does not eliminate the need for evaluation, validation, guardrails, and application-level controls.**

This is especially important for safety-related applications such as SIRA. SIRA should not make high-stakes safety decisions solely from an LLM response without appropriate human review and safeguards.


---

> **Key takeaway**

Prompt engineering is an **iterative engineering discipline**. The goal is NOT to write the longest prompt. The goal is to provide the **right instructions, context, examples, constraints, and output requirements** so the model can perform the intended task reliably.
