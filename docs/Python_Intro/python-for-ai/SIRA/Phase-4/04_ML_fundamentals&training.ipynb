### Phase 4

**Machine Learning Fundamentals & Model Training**

### Theme

> **Teaching the computer to recognize incident patterns**

Duration: **2 days**

---

### Learning objectives

By the end of this phase you should understand:

* What Machine Learning actually is
* Supervised vs Unsupervised Learning
* Why labels are important
* Features vs Target
* Training vs Testing
* Data Leakage
* Train/Test split
* Classification Algorithms
* Model Training
* Model Evaluation
* Saving models
* Loading models
* Making predictions

---

**Note**:

This is where you transition from **Data Scientists** to **Machine Learning Engineers**. Up until now, you have been building everything that comes **before** `machine learning`:

* Python
* Software Engineering
* OOP
* Project structure
* Data Loading
* EDA
* Data cleaning
* Feature Engineering

Now we finally arrive at the question you have been waiting for:

> **"Can we finally train an AI model?"**

Answer is `Yes`. But before writing

```python
model.fit(X_train, y_train)
```

you need to understand **what training actually means**. Most a times you immediately jump into `Scikit-learn`. But don't. Professional ML engineers first understand **what the model is learning**.
