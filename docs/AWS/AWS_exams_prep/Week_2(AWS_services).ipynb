### 2.1 What are Foundation Models?

Foundation models in AWS generative AI are large-scale machine learning models trained on vast, unlabeled data that serve as the base for building diverse artificial intelligence applications.

*In other words they are*:
- **Core base**: They act as the starting layer or foundation, meaning you do not have to train an AI model from scratch for every new task.
- **Self-Supervised Learning**: They use vast amounts of raw, unlabeled text, code, or images to learn general patterns instead of relying strictly on human-labeled data.
- **High versatility**: A single foundation model can perform many jobs, such as answering questions, writing code, summarizing text, or creating images.

**How AWS uses Foundation Models**

- *Amazon Bedrock* access*: AWS provides access to top foundation models through Amazon Bedrock, a managed service where you can use models from leading AI companies without managing any underlying servers.
- *Amazon Nova*: AWS offers its own family of high-performance foundation models called Nova to deliver fast text, image, and multimodal intelligence.
- *Customization*: You can tailor these base models to your specific business data using techniques like fine-tuning, prompt engineering, or retrieval-augmented generation (RAG).
- *Custom hardware*: AWS supports heavy model training and deployment using specialized cloud hardware chips called AWS Trainium and AWS Inferentia.

- **The Fix**: Spin up a tiny, cheap Amazon EC2 instance and let it run for 15-30 minutes. This passes basic account verification checks and often forces AWS backend systems to fully provision your account's access to Bedrock.

The next thing to test: Actual model invocation. Let's determine whether your account can actually invoke Nova 2 Lite from the same SageMaker environment.

Run this in a new Jupyter cell:
```python
import boto3
import json

bedrock_runtime = boto3.client(
    "bedrock-runtime",
    region_name="eu-central-1"
)

model_id = "amazon.nova-2-lite-v1:0"

request_body = {
    "schemaVersion": "messages-v1",
    "system": [
        {
            "text": "You are a helpful coding assistant that explains concepts step by step."
        }
    ],
    "messages": [
        {
            "role": "user",
            "content": [
                {
                    "text": "Explain the difference between Retrieval-Augmented Generation and tool use. Give a simple example of each."
                }
            ]
        }
    ],
    "inferenceConfig": {
        "maxTokens": 500,
        "temperature": 0.7
    }
}

response = bedrock_runtime.invoke_model(
    modelId=model_id,
    body=json.dumps(request_body),
    contentType="application/json",
    accept="application/json"
)

result = json.loads(response["body"].read())

print(json.dumps(result, indent=2))
```

### 2. Fine-Tuning vs. Prompt Engineering

These are two different ways to get a foundation model to do exactly what you want.

| Feature | Prompt Engineering | Fine-Tuning |
|---|---|---|
| What it is | Designing the perfect input text (instructions or examples) to guide the model's output. | Permanently updating the actual weights/parameters of the model using a custom dataset. |
| Analogy | Giving a smart intern a specific template and clear instructions before they write a report. | Sending that intern back to school for a specialized degree in your specific industry. |
| Data Needed | None to a few examples (few-shot prompting). | Hundreds or thousands of labeled, domain-specific training examples. |
| Effort & Cost | Very low and instantaneous. | Higher compute costs and requires data preparation time. |

### 3. Retrieval-Augmented Generation (RAG)
RAG is a framework that connects a foundation model to an external, live data source (like your company's internal wiki, database, or document repository) to provide accurate, up-to-date answers.

* The problem: Foundation models only know what they were trained on, resulting in a cutoff date or "hallucinations" when asked about specific internal files.
* The RAG solution: When a user asks a question, AWS searches your private data sources first to find the relevant documents. It then hands those documents to the model alongside the user's question.
* The Result: The model reads your documents and writes a perfectly summarized answer, complete with citations back to your source files.

### 5. Amazon SageMaker Clarify

> Is your AI fair?

*This is a question that every single person building a machine learning models has to be able to answer.*

Machine Learning models don't just invest bias out of thin air, it's like a mirror. They reflect the biases that are already out there in the world in the data we feed them. Sometimes they don't just reflect it they amplify it making it even worst sneaking into your model without you even realizing it and suddenly you got unfair or just plain wrong results. When we're building responsible AI, we're up against two things i.e `hidden bias` and `mysterious black box`.

This has real world consequencies. Imagine you're building an AI model thats deciding who get a loan, if the data it is trained on was all lopsided it might be super accurate for one group of people but terrible for another. Thats not just a bug in the code, thats a genuily unfair outcome that affects peoples lifes. So whats the solution here?

> That brings us to `Amazon SageMaker Clarify`.

Amazon SageMaker Clarify is a specialized feature within the [AWS SageMaker](https://aws.amazon.com/sagemaker/) machine learning platform designed to detect bias in datasets and models, provide model explainability, and generate compliance reports. It helps data scientists and ML engineers practice "Responsible AI" by making complex "black box" models transparent and auditable.

(*Note*: As of mid-2026, AWS has restricted SageMaker Clarify to existing customers only and is no longer onboarding new users to this specific feature as they transition toward open-source monitoring alternatives).

The easiest way to think about `Clarify` is like a set of diagnostic tools for your AI. It lets you pair insight to look at the data you're using and the model you built to make sure everything is working fairly. And it also helps you understand why it's working the way it is so you can actually explain it.

It works accross the entire ML workflow i.e 

- Bias detection before you train and after you train: `pre-train`.
- Interpreting what your model is actually doing.
- And monitoring it once its deployed in the real world.

**Step 1**: Bias Detection (Pre-Training, Post-Training, and Production)

Bias can sneak into machine learning pipelines at any stage. Clarify runs statistical checks across the entire lifecycle:

* Pre-Training Data: It scans raw datasets before training to catch data imbalances (e.g., checking if a financial loan dataset contains significantly fewer samples from a specific demographic or age group). 
* Post-Training Model: After training, it tests the model’s predictions to see if it acts unfairly (e.g., checking if the model accidentally approves credit at higher rates for one demographic group over another, even if their underlying financial metrics are identical). 
* In-Production Drift: Deployed models can degrade as real-world data shifts. Clarify integrates with SageMaker Model Monitor to trigger alerts if the model begins exhibiting biased patterns over time. 

**Step 2**: Model Explainability (SHAP Values)

Complex algorithms like Deep Neural Networks are highly accurate but incredibly difficult to understand. Clarify uses a game-theory algorithm called SHAP (SHapley Additive exPlanations) to explain individual or global predictions. 

* It calculates exactly how much weight each individual feature (e.g., credit history, annual income, or employment duration) contributed to the final output.
* It works beyond standard tabular data, offering explanations for Natural Language Processing (NLP/text) and Computer Vision (images).

**Step 3**: Governance and Compliance Reporting (Monitoring model in the real world feature attribution drift)

For highly regulated industries like banking, healthcare, and insurance, you cannot simply deploy an AI model without explaining why it makes certain decisions. Clarify automatically generates model governance reports. These are clean, visual summaries detailing data imbalances, fairness metrics, and feature importance scores that can be handed straight to internal risk, compliance, or external regulatory teams.

SageMaker model monitor is the scheduler i.e the tool that actually runs those clarify jobs over and over again on your life models keeping a constant eye on them for you.

### 6. Amazon Comprehend

Amazon Comprehend is a fully managed Natural Language Processing (NLP) and text-analytics service provided by Amazon Web Services (AWS). It uses machine learning to automatically inspect and extract structural insights, sentiment, and data from unstructured text—like emails, social media feeds, customer support tickets, and call logs—without requiring you to build or train your own ML models. [1, 2, 3] 
Instead of building complex algorithms from scratch, developers can simply make a standard API call to Comprehend, send a block of text, and receive structured JSON responses back.

**Core features of Amazon Comprehend**
Amazon Comprehend includes several pre-trained out-of-the-box text analysis tools:

* Sentiment Analysis: Identifies the overall emotional tone of the text. It classifies strings as Positive, Negative, Neutral, or Mixed with a corresponding confidence score.
* Entity Recognition: Scans text to identify and categorize specific items such as people, places, dates, brands, and organizations.
* PII Detection and Redaction: Automatically detects Personally Identifiable Information (PII) like credit card numbers, email addresses, phone numbers, and SSNs. It allows you to automatically redact or mask this sensitive data to ensure privacy compliance.
* Key Phrase Extraction: Automatically detects the main talking points or nouns in a document (e.g., extracting "excellent customer service" or "broken charging port" from a product review).
* Language Detection: Automatically identifies the dominant language spoken across over 100+ supported languages.
* Syntax Analysis: Breaks down sentences into grammatical parts of speech, distinguishing between nouns, verbs, adjectives, and punctuation.

**Custom and advanced capabilities**

If the pre-trained models don't fit your business needs, the service can expand into custom territory:

* Custom Classification: You can upload your own training text to teach Comprehend to tag documents with unique categories specific to your company (e.g., routing support emails straight to "Billing," "Technical Support," or "Legal").
* Custom Entity Recognition: Teaches Comprehend to locate domain-specific words unique to your industry, like specific product SKU numbers, medical terms, or financial codes.
* Intelligent Document Processing (IDP): Comprehend can process data straight from PDFs, Word documents, and images without needing you to strip out the text first.

**Important architecture shifts**
As LLMs (Large Language Models) have evolved, AWS has shifted its strategy regarding certain legacy text-processing tools.
As of April 30, 2026, features like Topic Modeling, Event Detection, and Prompt Safety Classification entered maintenance mode and are no longer available to new customers. For these tasks, AWS now redirects users to handle them using generative AI via [Amazon Bedrock](https://aws.amazon.com/bedrock/) LLMs and Bedrock Guardrails.

**Example real-world use case**

A company routes all phone transcripts from their call center directly into an [Amazon S3](https://aws.amazon.com/s3/) bucket. An automated system sends that text to Amazon Comprehend, which automatically masks out customer credit card numbers (PII), tags the log as "Negative Sentiment" if the user was upset, and extracts key phrases like "billing error" so the company can track trending complaints in real-time.
