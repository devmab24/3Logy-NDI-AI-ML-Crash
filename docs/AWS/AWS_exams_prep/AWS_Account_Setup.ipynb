## Setting up AWS Account

> **Just a warning that whenever you create an `AWS` account** if you don't have one already, it'll ask for your credit card details, it's super save. So you can add it and also note that the first time you setup the account they might take $1 from your credit card but you don't have to worry about that because they just do that to verify that it's a legit credit card but once it's verify they'll actually revert that $1 back so its going to cause you $0.

**Step by step procedure**

* On your `chrome` browser search menu type: `AWS` and click search.
* Select or click on `Amazon Web Services` which appears at the top.
* Note: it may likely redirect you to some other aws page, if so `click the page logo` to go back to to main page or you may navigate to: `aws.amazon.com`.
* Find the `Sign-In` button at the top right corner of the page and click. 
* You may have different screen but dont worry about that.
* You'll notice that there's a `Root` user and `IAM useer` Sign-In

The difference between `Root` user and `IAM useer` is the following.

* When you first create your AWS Account, you're going to have/be the `Root` user.
* And the `Root` user is the full `Admin` i.e you're the entire administrator of your account and you can do anything.
* And that's why you should be very careful with using your `Root` user Account to Sign-In, because if someone gets the credentials to your `Root` user account you'll likely have to close your AWS Account and we don't want that because they're going to have access to all the tools, they can actually lock you out of your own account. So it's very dangerous.
* So, that's why it's a very good practice to `Sign-In` to your `Root` user Account but from there you start creating `IAM users` to use.
* And the reason this is good is because, if a `hacker` gets your credentials to your `IAM user` then you can just Log In to your `Root` user Account and `delete` that `IAM user`. So it's not a big deal.
 
So, I want you to get in the habbit as to whenever you're using `AWS` to `NOT` Sign-In with your `Root` user. 

And again if this is your first time Signing up to AWS then you'd have to `Sign In` or `Sign Up` with the `Root` user.

### To `Sign Up`

**Click the `Create a new AWS Account` button at the bottom and complete the registration form.**

After a succesful login you'll be redirected to your `dashboard`

> Now that you've Logged In to your Account, the next step is to setup an `IAM` user and with that `IAM` user we can LogIn to thesame `AWS` Account but not using the `Root` user to LogIn.

## Setting up `IAM` Roles + Best Practices

If you don't see the `IAM` as a recently visited, click the search input menu and search for `IAM` i.e `Identity Access Management`.
* From the left menu list, click on `Users`.
* You'll see the `Create user` button at the top right corner. Click on it.
* Specify the user details. Fill in your `username`. It's doest have to be a global name, you may name it whatever you want.
* Tick the box below it to give it access to the Console which is what you see i.e the user interface otherwise you only interact using the terminal.
* Select to create an `IAM user` as the `user` type. Don't worry with the recommended option as that will be used mostly when you start working with a company where the IT will set that up for you. Hence; you'll never have to do this kind of thing.
* Specify the `password`. Tick the `autogenerate` password and also select that users must create a new password at next `sign-in`.
* Setup the permissions option. Select the `Attach policies directly` since we don't have current groups that we're working with.
* Search and Tick the `AdministratorAccess` for the `IAM` user. (Although this is bad practice but because we'll need so much services thats why we'll give the `AdministratorAccess` to the `IAM` user otherwise don't do that.)
* Click next and then finally, click on the `Create User` button.
* Now the `IAM` user has been created you may download the `csv file` or copy the URL and password directly.
* Now, if you return to the Users page you'll see the newly created user there.

**Note**: In productions accounts, DevOps or the IT/Security may setup some rules that you've to change your password every 3 months or MFA has/must be enabled etc.

## AWS Security Best Practices

> LogIn using the credentials you created earlier and make sure to change the password

> Always remember to use the same **region**

> In case of a breach or your Account got hacked, how do you delete the user?

* From the `Root` user Click on the users list.
* Select the user to be `deleted`
* Click the `delete` button at the top right corner.
* Confirm the deletion
* Delete the user 

To select or change **region**, Click on the region dropdown menu and choose the closest to you.
