## The goal is to:

> Pass the AWS AI Practitioner exam while understanding enough practical AI on AWS to answer scenario-based questions confidently.


**The exam heavily focuses on**:

- AI/ML fundamentals

- Generative AI

- Foundation Models (FMs)

- Amazon Bedrock

- Responsible AI

- Security & Governance

- Business use cases

- AI lifecycle

- AWS AI services


with less emphasis on coding and deep mathematics.
