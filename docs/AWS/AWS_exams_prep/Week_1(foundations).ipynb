### What is artificial intelligence?

Definition:

AI is the ability of computers to perform tasks that normally require human intelligence.

Examples:

* Speech recognition
* Chatbots
* Recommendation systems
* Fraud detection
* Self-driving vehicles

---

### What is machine learning?

Definition:

Machine Learning is a subset of AI where systems learn patterns from data without being explicitly programmed.

Example:

Instead of writing:

```python
if email_contains("win money"):
    spam = True
```

the model learns spam patterns from historical emails.

---

### What is deep learning?

Definition:

Deep Learning uses neural networks with multiple layers to learn complex patterns.

Examples:

* Face recognition
* Image classification
* LLMs
* Voice assistants

## Types of machine learning

---

### 1. Supervised Learning

Learns from labeled data.

Example:

| Email            | Label    |
| ---------------- | -------- |
| Win Money        | Spam     |
| Meeting Reminder | Not Spam |

Applications:

* Spam Detection
* Credit Scoring
* Medical Diagnosis

---

### 2. Unsupervised Learning

Learns patterns without labels.

Applications:

* Customer Segmentation
* Market Basket Analysis
* Anomaly Detection

---

### 3. Reinforcement Learning

Learns using rewards and penalties.

Applications:

* Robotics
* Game Playing
* Autonomous Vehicles

## Foundation models (FMs): Definition

Large models trained on vast datasets that can be adapted for many tasks.

Examples:

* Claude
* Llama
* Amazon Nova
* GPT

---

### Traditional ML vs Foundation Models

| Traditional ML     | Foundation Model        |
| ------------------ | ----------------------- |
| Specific task      | Multiple tasks          |
| Smaller data       | Massive data            |
| Easier explanation | More complex            |
| Faster training    | Very expensive training |

AWS recently added objectives comparing when to use traditional ML vs Foundation Models.

## Training vs Inference:

### Training

When the model learns patterns.

Example:

```text
Historical Data
      ↓
Model Learns
```

---

### Inference

When the trained model makes predictions.

Example:

```text
New Email
    ↓
Spam Model
    ↓
Prediction
```
