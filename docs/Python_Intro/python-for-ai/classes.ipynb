### What is `self`?

In `Python`, `self` is not a keyword but a convention. `self` refers to the current instance of the `Class (object)` being created.

Whenever an object calls a method, Python automatically passes that object as the first argument.

Think of `self` as:

- "This specific object."

`````python
class Participant:

    def greet(self):
        print(f"Welcome, {self.name}!")

p = Participant() #You may also pass the name as parameter if positional arguments is needed
p2 = Participant()

p.name = "Danladi"
p.age = 20
p.score = 90
p2.name = "Danladi2"
p2.age = 25
p2.score = 85
p.greet()
p2.greet()
`````

When we call an instance method like this, Python does something behind the scenes. It automatically passes the instance as the first argument to the method, this means `self` in a method is essentially this varable. In our example above we added a name attribute to `p` dynamically and `self.name` let's the method access it, without`self` Python wouldn't know which object name to look at.

We know `Python` is flexible so we can keep adding attributes to the objects like; `p.age = 22, p.id = 001, etc` but while this works it can get messy really quickly. Imagine if we have to add `5` or `10` attributes everytime we created an object, this where the `__init__` method saves us. It helps us keep our code clean.

Think of `Attributes` as a variable that belongs to a class or an object, used to store data that describes its state or characteristics. They're called `properties` or `features` of a class and you can access them using `dot notation`.

`````python
class Cat:

    def __init__(self, name, age):
        self.name = name
        self.age = age

#Creating an instance of the Dog class; Note: the init method is automatically called and it takes the parameters
#name and age and assigns them as attributes to the object `my_cat()`
my_cat = Cat("Leo", 2)
my_cat2 = Cat("Milo", 3)

print(my_cat.name)
print(my_cat.age)
print(my_cat2.name)
print(my_cat2.age)
`````

`````python
class MLModel:
    """Simulates training the model on data."""
    def __init__(self, learning_rate=0.01, max_iter=100):
        # Sets up the hyperparameters and initial state
        self.learning_rate = learning_rate
        self.max_iter = max_iter
        self.is_trained = False
        self.weights = None

    def fit(self, X_train, y_train):
        """Method 1: Simulates training the model on data."""
        print(f"Training model with learning rate {self.learning_rate}...")
        
        # Simulate weight optimization loop
        self.weights = [0.5, -1.2, 3.1]
        self.is_trained = True
        
        print(f"Training complete after {self.max_iter} iterations.")

    def predict(self, X_new):
        """Method 2: Makes predictions using the trained weights."""
        if not self.is_trained:
            print("Error: Cannot predict. Train the model first using .fit()!")
            return None
            
        print(f"Processing {len(X_new)} new samples using weights {self.weights}...")
        # Simulate generating a prediction list
        predictions = [1, 0, 1] 
        return predictions
`````

`````python
# 1. Initialize the model with custom hyperparameters
my_model = MLModel(learning_rate=0.05, max_iter=500)
my_model.fit([[1, 2], [3, 4], [5, 6]], [1, 0, 1])  # Train the model with dummy data
# # 2. Try to predict before training (Triggers the safety check)
# my_model.predict([[1, 2, 3]]) 

# # 3. Train the model using the fit method
# dummy_X = [[1, 2], [3, 4], [5, 6]]
# dummy_y = [1, 0, 1]
# my_model.fit(dummy_X, dummy_y)

# # 4. Successfully make predictions now that it is trained
results = my_model.predict([[7, 8], [9, 10], [11, 12]])
print(f"Predicted outputs: {results}")
`````

### Working with Lists and Dictionaries in Classes

### Combining Lists and Dictionaries

Example: A school can have:

- Multiple students

- Each student has:

    - Multiple subjects and scores

`````python

#Create the Class
class Classroom:

    def __init__(self):

        self.students = []

    def add_student(self, name, scores):

        student = {

            "name": name,

            "scores": scores

        }

        self.students.append(student)

    def display_students(self):

        for student in self.students:

            print("\nStudent:", student["name"])

            for subject, score in student["scores"].items():

                print(subject, ":", score)

## Create Object
classroom = Classroom()

classroom.add_student(
    "John",
    {
        "Python": 90,
        "ML": 80
    }
)

classroom.add_student(
    "Mary",
    {
        "Python": 95,
        "ML": 88
    }
)

classroom.display_students()
`````

## Inheritance & Polymorphism

Using inheritance we can reuse parent class code. Inheritance allows us to define a class that inherits all the methods and properties from parent class. The parent class or super or base class is the class which gives all the methods and properties. Child class is the class that inherits from another or parent class.
Let us create a student class by inheriting from person class.

`````python
class Person4:
      city = "ArithmeticError"
      
      def __init__(self, firstname='Adam', lastname='Yahya', age=20, country='Nigeria', city='Gombe'):
          self.firstname = firstname
          self.lastname = lastname
          self.age = age
          self.country = country
          self.city = city
          self.skills = []

      def person_info(self):
        return f'{self.firstname} {self.lastname} is {self.age} years old. He lives in {self.city}, {self.country}.'
    
      def add_skill(self, skill):
          self.skills.append(skill)

p1 = Person4()
# print(p1.person_info())
# p1.add_skill('HTML')
# p1.add_skill('CSS')
# p1.add_skill('JavaScript')

# p2 = Person4('Silas', 'Iliya', 30, 'Nigeria', 'Bauchi')
# print(p2.person_info())
# print(p1.skills)
# print(p2.skills)
`````

`````python
class PersonChild(Person4):
    pass

s1 = PersonChild('Eyob', 'John', 30, 'Nigeria', 'Abuja')
s2 = PersonChild('Lidiya', 'Yelwata', 28, 'Nigeria', 'Kaduna')
print(s1.person_info())
s1.add_skill('JavaScript')
s1.add_skill('React')
s1.add_skill('Python')
print(s1.skills)

print(s2.person_info())
s2.add_skill('Organizing')
s2.add_skill('Marketing')
s2.add_skill('Digital Marketing')
print(s2.skills)
`````

`NOTE`: We did not call the **__init__**() constructor in the child class. If we didn't call it then we can still access all the properties from the parent. But if we do call the constructor we can access the parent properties by calling _super_.  
We can add a new method to the child or we can override the parent class methods by creating the same method name in the child class. When we add the **__init__**() function, the child class will no longer inherit the parent's **__init__**() function.

### Overriding parent method

`````python
class StudentChild2(Person3):
    def __init__ (self, firstname='Bilkisu', lastname='Sani',age=250, country='Morocco', city='Marakesh', gender='female'):
        self.gender = gender
        super().__init__(firstname, lastname,age, country, city)
        
    def person_info(self):
        gender = 'He' if self.gender =='male' else 'She'
        
        return f'{self.firstname} {self.lastname} is {self.age} years old. {gender} lives in {self.city}, {self.country}.'

s1 = StudentChild2()
s2 = StudentChild2('Lidiya', 'Amos', 28, 'Finland', 'Espoo', 'female')

print(s1.person_info())
s1.add_skill('JavaScript')
s1.add_skill('React')
s1.add_skill('Python')
print(s1.skills)
print()
print(s2.person_info())
s2.add_skill('Organizing')
s2.add_skill('Marketing')
s2.add_skill('Digital Marketing')
print(s2.skills)
`````

### Class Attributes vs Instance Attributes

### Why are we learning this?

When building applications, not all data belongs to a specific object. Sometimes information should be shared across all objects.

For example:

- Student A attends AI Class.

- Student B attends AI Class.

- Student C attends AI Class.

The class name is the same for everyone.

It doesn't make sense to store:

class_name = "AI Class"

inside every student object.

Instead, we can store it once at the class level.

This introduces us to:

- Class Attributes
- Instance Attributes

### What is an Instance Attribute?

An instance attribute belongs to a specific object. Each object gets its own copy.

Example:

Student A:

Name = John

Student B:

Name = Mary

Student C:

Name = Ahmed

Every student has a different name.

Therefore:

name should be an Instance Attribute.

`````python
class Students:
    """
    When we create:
        student1 = Student("John")
    Python stores:
        student1.name = "John"

    When we create:
        student2 = Student("Mary")
    Python stores:
        student2.name = "Mary"

    Each object has its own value.

    These are called: Instance Attributes
    """
    def __init__(self, name):

        self.name = name

student1 = Students("John")
student2 = Students("Mary")

print(student1.name)
print(student2.name)
`````

### What is a Class Attribute?

A class attribute belongs to the class itself. All objects share the same value. It is defined outside the constructor.

`````python
class Student:

    school = "3Logy Academy"

    def __init__(self, name):

        self.name = name

student1 = Student("John")
student2 = Student("Mary")

print(student1.school)
print(student2.school)
`````

### Why use Class Attributes?

Because the value is shared. Instead of storing:

school = "3Logy Academy"

inside every student, Python stores it once and allows every object to access it.

This saves memory and keeps code cleaner.

`````python
class Student:

    school = "3Logy Academy"

    def __init__(self, name):

        self.name = name

student1 = Student("John")
student2 = Student("Mary")

print(Student.school)
`````

### Comparing class and instance attributes

| Class Attribute | Instance Attribute |
|---------------|--------------------|
| Shared by all objects | Unique to each object |
| Defined outside __init__ | Defined inside __init__ |
| One copy exists | Every object gets a copy |
| Accessed through class or object | Usually accessed through object |

`````python
class Student:

    school = "3Logy Academy"

    def __init__(self, name):

        self.name = name

student1 = Student("John")

print(Student.school)
print(student1.name)
`````

### Step 4: Abstraction

## 1. What is Abstraction? (The Analogies)

Abstraction means: `Showing only essential features while hiding implementation details` i.e it means hiding the messy, complicated details and showing only the essential parts.

* The Car Analogy: When you drive a car, you only use the steering wheel, gas pedal, and brakes. You do not need to understand how the engine burns fuel or how the transmission shifts gears to drive to the store. The steering wheel is the abstract interface that hides the engine's complexity.

`AI Example` Every AI model should have:
- `train()`
- `predict()`
- `save_model()`

But each model implements them differently.

* The TV Remote Analogy: To change the channel, you press a button. You do not need to know how the remote converts that press into infrared light or how the TV decodes the signal.

In programming, abstraction allows us to use code without needing to understand every single line hidden inside it.

------------------------------

### 2. How Python Does It: The Template

Python does not do abstraction automatically. We have to import a built-in tool called abc (Abstract Base Classes).
Think of an Abstract Class as a strict blueprint or a rulebook.

   1. Abstract Class (The Blueprint): A class that cannot be used to make objects directly. It only exists to be inherited by other classes.
   2. Abstract Method (The Rule): A generic method inside the blueprint that has no code inside it. It forces any child class to create its own version of that method.

------------------------------

### Key Rules to Remember

* Rule 1: You cannot create an object directly from an Abstract Class (e.g., Animal() is forbidden).
* Rule 2: If a child class (like Dog) forgets to write the abstract method (make_sound), Python will throw an error and refuse to run the program.

------------------------------
### Why do we use Abstraction?

* It avoids mistakes: It acts as a safety checklist. If a teammate creates a new Cow class, the blueprint forces them to remember to add a make_sound method.
* It simplifies teamwork: You only need to know what a class does, not how it does it. You can use a classmate's code just by knowing the method names.
* Simplicity and clarity (Readability)
* Modularity and Reusability
* Encapsulation of Comlexity
* Flexibility and Extensibility

### When to use it:

- Dealing with comlex systems where there are many interrelated components.
- Designing libraries that'll be used by others, as it helps in creating clear APIs.
- Promoting code maintainability and scalability, especially in large projects where clarity and modularity are crucial.

---

`Python` has a `build-in` module named `abc` i.e `Abstract, Base, Class` and `abstractmethod` decorator to create an abstract interface.

Note: there are certain property to keep in mind which are:
- You cannot create object of the Abstract class i.e obj1 = AIModel()
- Abstract method should be a part of the abstract class
- If you want to create abstract method, you cannot define it

The `syntax` below shows how to create an abstract class

```
class AIModel(ABC):
    """Syntax"""
    @abstractmethod
    def train(self):
        pass

    @abstractmethod
    def predict(self):
        pass

    @abstractmethod
    def save_model(self):
        pass
```

`````python
# import ABC
from abc import ABC, abstractmethod

#Let's create a class `Vehicle`
class Vehicle(ABC):
    """Abstraction"""
    def __init__(self, make, model, year):
        #Every vehicle must have these properties
        self.make = make
        self.model = model
        self.year = year
    
    @abstractmethod #this tells `python` whatever comes after this i.e the method, will be an abstract method.It's just a reminder.
    def display_info(self):
        """Now, whatever class that takes our base class `Vehicle` must complete this method."""
        pass


#Let's create a class `Vehicle`
class Car(Vehicle):
    """Abstraction"""
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors
        
    
    def display_info(self):
        """Now here we must complete the `display_info()` because we inherit Vehicle which is an ABC, we must complete it."""
        print(f"Car: {self.year}, {self.model}, {self.num_doors} doors")

#Using the Abstraction
car = Car("Porsche", "Cayene S", 2022, 4)
car.display_info()
`````

In summary, we create an `Abstract Base Class` as a reminder, so any class that inherits it, it does not care what you do but `you must complete the functions that the ABC offers`.

Think of a class structure for managing students and enrollments, let's take `Mathematics` as the subject. You may have a `Course` as the base class that reminds child classes all nethods that must be completed for each individual student course. 

`````python
# import ABC
from abc import ABC, abstractmethod

#Lets take a look at the templates below.

#This could become an abstract base class
class Course:
    """abstract base class"""
    def __init__(self, course_code, course_name):
        self.course_code = course_code
        self.course_name = course_name
        
    #these 3 methods we're not going to complete them here because they're `Abstract methods`, we're only going to indicate.
    @abstractmethod
    def enroll_student(self):
        """Abstract"""
        pass
    
    @abstractmethod
    def withdraw_student(self):
        """Abstract"""
        pass

    @abstractmethod
    def calculate_grade(self):
        """Abstract"""
        pass

#We must complete the methods for the Math class since it'll inherits the Course class which is an ABC.
class MathCourse(Course):
    def __init__(self, course_code, course_name):
        super().__init__(course_code, course_name)
        self.students = {}
    
    def enroll_student(self, student):
        self.students[student.student_id] = {"name": student.name, "grades": []}
    
    def withdraw_student(self, student):
        del self.students[student.student_id]

    def calculate_grade(self, student):
        if student.student_id in self.students:
            grade = sum(self.students[student.student_id]["grades"]) / len(self.students[student.student_id]["grades"])
            return f"Average grade for {student.name}: {grade}"
        else:
            return f"Student not enrolled in this Course."
    
#We'll have lots of students, so we may need to store them as lists. Example student_info like ID, and NAME.
class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
    
    def submit_assignment(self, course, grade):
        course.students[self.student_id]["grades"].append(grade)
        print("Grade Added Successfully.")
    
`````

`````python
#Let's use the code
math = MathCourse("Linear Algebrae-101", "Matrix Intro")

student1 = Student("002", "Adam")
student2 = Student("005", "Maryam")

math.enroll_student(student1)
math.enroll_student(student2)

student2.submit_assignment(math, 90)
student2.submit_assignment(math, 85)
student1.submit_assignment(math, 76)
student1.submit_assignment(math, 100)

print(math.calculate_grade(student1))
print(math.calculate_grade(student2))
`````

The error occurs because Python's abc (Abstract Base Class) module prevents you from creating an instance of a subclass if it fails to implement all abstract methods defined in its parent class.In our Stripe class, we have a typo: we named the method `pa` instead of `pay`. Because pay is missing, Python considers Stripe to still be an abstract class.

`The Fix`

Change def pa(self): to def pay(self): in your Stripe class.Here is the corrected code:

`````python
from abc import ABC, abstractmethod

class PaymentGateway2(ABC):
    """Abstraction"""
    @abstractmethod
    def pay(self):
        pass


class PayPal2(PaymentGateway2):
    """Abstraction"""
    def pay(self):
        print("paying using Paypal...")
        

class Stripe2(PaymentGateway2):
    """Abstraction"""
    # FIXED: Changed 'pa' to 'pay'
    def pay(self):
        print("paying using Stripe...")
    
    
class Purchase2:
    """Abstraction"""
    def __init__(self, gateway):
        self.gateway = gateway

    def checkout(self):        
        self.gateway.pay()
        
        print("Checking out...")

gateway1 = PayPal2()
gateway2 = Stripe2()  # This will now work without errors

purchase = Purchase2(gateway2)
purchase.checkout()
`````

### Let's combine all concepts in a single example

`````python
from abc import ABC, abstractmethod

# ABSTRACT CLASS
class AIModel(ABC):

    @abstractmethod
    def train(self):
        pass

    @abstractmethod
    def predict(self):
        pass

    @abstractmethod
    def save_model(self):
        pass

#INHERITANCE
# CHILD CLASS 1
class ClassificationModel(AIModel):

    def __init__(self):

        self.__accuracy = 0

    def set_accuracy(self, accuracy):

        if 0 <= accuracy <= 100:

            self.__accuracy = accuracy

    def get_accuracy(self):

        return self.__accuracy

    def train(self):

        print("Training Classification Model")

    def predict(self):

        print("Predicting Classes")

    def save_model(self):

        print("Classification Model Saved")


# CHILD CLASS 2
class RegressionModel(AIModel):

    def train(self):

        print("Training Regression Model")

    def predict(self):

        print("Predicting Numerical Values")

    def save_model(self):

        print("Regression Model Saved")


# OBJECTS
classifier = ClassificationModel()

regressor = RegressionModel()

classifier.set_accuracy(94)


# POLYMORPHISM
models = [

    classifier,

    regressor

]

for model in models:
    model.train()
    model.predict()
    model.save_model()

    print()
`````
