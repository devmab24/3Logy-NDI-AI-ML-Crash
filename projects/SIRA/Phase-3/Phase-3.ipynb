### Build `DataCleaner`

Instead of many functions use a class.

```python
class TextPreprocessor:
```

---

```python
import re
import string


class TextPreprocessor:

    def lowercase(self, text):
        return text.lower()

    def remove_punctuation(self, text):
        return text.translate(
            str.maketrans("", "", string.punctuation)
        )

    def remove_numbers(self, text):
        return re.sub(r"\d+", "", text)

    def remove_spaces(self, text):
        return " ".join(text.split())
```

---

```python
processor = TextPreprocessor()

text = "Pressure leak detected on Pipeline 7!!!!"

text = processor.lowercase(text)

text = processor.remove_numbers(text)

text = processor.remove_punctuation(text)

text = processor.remove_spaces(text)

print(text)
```

### `Stop Words`

Explanation:

Words like

```text
the

is

are

an

a

of

to

on
```

appear everywhere. They rarely help classification.

---

Example

```text
pressure leak detected on pipeline
```

---

Using NLTK

```python
from nltk.corpus import stopwords
```

---

Method

```python
def remove_stopwords(self,tokens):

    stops = set(stopwords.words("english"))

    return [
        word
        for word in tokens
        if word not in stops
    ]
```

### Create feature_engineering.py

Instead of `main.py` again professional engineering.

```python
from sklearn.feature_extraction.text import TfidfVectorizer


class FeatureEngineer:

    def __init__(self):

        self.vectorizer = TfidfVectorizer()

    def fit_transform(self,texts):

        return self.vectorizer.fit_transform(texts)

    def transform(self,texts):

        return self.vectorizer.transform(texts)
```
