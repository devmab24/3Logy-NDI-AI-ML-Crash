### The goal is to:

> Pass the AWS AI Practitioner exam while understanding enough practical AI on AWS to answer scenario-based questions confidently.


**The exam heavily focuses on the below domains**:

* Domain 1: Fundamentals of AI and ML                             - 20%

* Domain 2: Fundamentals of Generative AI (Amazon Bedrock)        - 24%

* Domain 3: Application of Foundation Models (FMs)                - 28%

* Domain 4: Guidelines for Responsible AI                         - 14%

* Domain 5: Security, Compliance & Governance for AI Solutions    - 14%

* Others: Business use cases, AI lifecycle, AWS AI services



> Note: In the exams there's less emphasis on coding and deep mathematics.

## Recommended tools

1. **AWS skill builder (mandatory)**: The exam is designed around AWS concepts. You should create accounts on: [AWS Skill Builder](https://skillbuilder.aws?utm_source=chatgpt.com)

**Use**:

- AI practitioner learning path

- Practice questions

- Exam-style assessments

---

2. **AWS free tier**: [AWS Free Tier](https://aws.amazon.com/free?utm_source=chatgpt.com)

**This allows exposure to**:

- AWS Console, IAM, S3, SageMaker Studio, Bedrock (where available)


This enables you to have familiarity with the AWS environment.

---

3. **SageMaker studio**: [Amazon SageMaker AI](https://aws.amazon.com/sagemaker/?utm_source=chatgpt.com)

**It's used to demonstrate**:

- Dataset upload, Notebooks, Model training concepts, JumpStart and Model deployment concepts


This demonstration session is valuable because it expose you to SageMaker studio.

---

4. **PartyRock**: [Amazon PartyRock](https://partyrock.aws/?utm_source=chatgpt.com)

**Excellent for**:

- Prompt engineering, Generative AI concepts, Foundation models, AI applications

This lets you build AI apps without coding.

---

5. **Amazon bedrock demonstrations**: [Amazon Bedrock](https://aws.amazon.com/bedrock/?utm_source=chatgpt.com)

**The exam increasingly emphasizes**:

- Foundation Models, GenAI, Agents, Prompt engineering, RAG concepts, and Evaluation

These appear repeatedly in the latest guide.
