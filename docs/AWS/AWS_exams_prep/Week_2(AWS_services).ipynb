### 1. **SageMaker**

The `Amazon SageMaker AI` (formerly known simply as Amazon SageMaker) is a tool used prepare, to train, tune, build, manage and deploy machine learning models. 

Amazon SageMaker AI is a fully managed cloud service i.e an entire collections of things that enables developers (AI/ML) to build, train, and deploy ML and AI models at scale. It's got `toolkits`, `SDKs`, `APIs`, `example code`, `containers`, entire tools you can use like sample code, sample project, documentation etc.

> Note: Their's also the `AWS SageMaker` which is another service that's specifically used for data analytics and AI.

`SageMaker` has layers which is listed below: 

* User Interface
* Machine Interface
* Infrastructure

* 1. **User Interface**: Consists of three distinct categories which are:- 
        * **AWS Console**: This is the user interface/UI we use to do all the AWS services like training and deploying models from the web.
        * **SageMaker Notebooks**: Is a managed jupyter notebook server. They allow you to interact and work with specifically machine learning and data science tasks.
        * **SageMaker Studio**: The studio is actually a sort of iteration on top of jupyter notebooks i.e a `Lab` opensource projects that has `extensions` and `adons` preinstalled into it that help it to integrate with sagemaker services. It has a notebook centric as well but with additional tools that provides extra capability.

* 2. **Machine Interface**: These allows us to interact with the services underneath, and they include:-
        * **API**: Application Programming interface is used to communicate and connect into to be able to configure all the services in the AWS e.g create S3 buckets, create EC2 instances and also it allows you to create and manage your resources inside of sagemaker. But we don't usually use the API, most a times we used the `console` which uses the API under the hood or we use the SDK. 
        * **SDK**: Software Development Kits (`SDK`) interacts with the `API`  on our behalf, it abstracts us away from the complexity of dealing with the API and puts us in a more familiar development environment. One of the popular choice for me is the `Boto3` since we're using `Python` but there are others available for other languages as well.
        * **SageMaker SDK**: These SDK works thesame but they add a little bit of business logic that help you work with sagemaker which is importance inside of the whole amazon environment.

* 3. **Infrastructure**: This infrastructure level is what gives you all machines thats required for your business operations. They include:-
        * **SageMaker Containers**: They're managed by AWS to integrate into the sageMaker environment. They form the important part of the way sageMaker works. The containers have setups (not much though) in them i.e the running libraries e.g pytorch, tensorflow, scikit-learn, ms-net etc with the capability inside of them to run the workloads that you provide. So you don't need to deal with these containers unless you want to, you just put your code and run them. You don't need to actually orchestrate the containers as well.
        * **Built-in Algorithms (Containers)**: These are fully managed containers that actually have `algorithms` already on them. So they're  built-in algorithms into sagemaker which can solve number of the problems that you might have in common machine learning problems straight away e.g Classification, CNN, Object Detection, PCA, XGBoost etc they're put together and put inside of a container and optimized  to run inside the AWS environment.
        * **Containers (Orchestration)**: This gives you the ability to orchestrate your own container if you want to.

> A SageMaker AI **notebook instance** is a fully managed cloud computer (virtual machine) that comes pre-installed with Jupyter Notebook applications and machine learning libraries. It gives data scientists, and ML Engineers a ready-to-use environment to write code, explore data, and build models without setting up any local software.

Steps to create a **Jupyter Notebook**:

* From the Console, type `SageMaker AI` in the search bar and click on the search button.
* From the left side menu lists:
    * Find `Applications and IDEs` list.
    * Select **Notebooks**.
    * Select the `Notebooks Instances` tab. .
* On the Console/UI Click  on `Create notebook instance`

> Note: When you click "Create notebook instance" in Amazon SageMaker, AWS handles a multi-step orchestration process behind the scenes. It sets up the physical hardware, configures the networking, loads the machine learning software, and establishes security bridges.

**Here is the exact step-by-step breakdown of what happens under the hood:**

* **Spins up a server**: AWS launches a dedicated virtual machine (EC2 instance) based on the size you chose.
* **Attaches a hard drive**: It links a persistent storage volume (EBS) to save your files when the machine is turned off.
* **Loads the software**: It runs a container pre-packaged with Linux, JupyterLab, and AI libraries like PyTorch and TensorFlow.
* **Secures the network**: It applies your IAM permissions so you can securely access Amazon S3 data without typing passwords.
* **Creates a secure link**: It generates an encrypted web URL so you can log into your workspace from your browser.

### **Quota Increase**

When you create a `SageMaker` resource, such as a processing job, training job, endpoint, or Studio app, you might get the **ResourceLimitExceed** error. See the following example error message:

> The account-level service limit ml.m5.xlarge for endpoint usage is 0 instances, with current utilization of 0 Instances and a request delta of 1 instances. Please contact AWS support to request an instance for this limit

This error usually occurs when you exceed the account level service quotas that are specified for your SageMaker resources.

What this means for you?

If you're new to AWS and you've just setup an account, sometimes you'll NOT have access to every single resource immediately and the reason for that is because AWS wants to protect you, they don't want you to accidentally spin-up a GPU instance since you're a new user and you might be making a mistake, So instead they will not allow you to have a GPU instance, and this is the kind of error message that you might encounter. You might encounter this error message throughout the course `three times` or more. `When we setup a training job`, `When we deploy an endpoint`, or `When we setup Load testing for endpoints`.

We have not discuss any of these yet, but this should give you an idea of what the error message actually is. To fix this, go to:
* AWS Console and type `Quotas`.
* Click on `Service Quotas`.
* At the top right find `Manage quotas` and search for `Amazon SageMaker`.
* Click `View Quotas`.
* In the search bar, type `the instance type` e.g `ml.p2.xlarge instance for training`.
* Ask for an increase by selecting the instance then click the `Request increase at account level` button at the top right.
* Type in the number of quota and click `Request`.

During Training we'll use `ml.p2.xlarge for training job usage`, so by default aws sets the quota to 0 to protect you because instances costs money.
