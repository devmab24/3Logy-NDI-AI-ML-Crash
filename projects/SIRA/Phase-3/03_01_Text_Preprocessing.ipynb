### Build `DataCleaner`

Instead of many functions use a class.

```python

```

---

```python
import re
import string


class TextPreprocessor:

    def lowercase(self, text):
        return text.lower()

    def remove_punctuation(self, text):
        return text.translate(
            str.maketrans("", "", string.punctuation)
        )

    def remove_numbers(self, text):
        return re.sub(r"\d+", "", text)

    def remove_spaces(self, text):
        return " ".join(text.split())
```

Using NLTK

```python

```

---

Method

```python
def remove_stopwords(self,tokens):

    stops = set(stopwords.words("english"))

    return [
        word
        for word in tokens
        if word not in stops
    ]
```

### Lemmatization

Difference is that `stemming` cuts words. `Lemmatization` understands grammar.

Example

```text
better

↓

good
```

Possible with `WordNet`.

Explanation:

Professional NLP usually prefers lemmatization.

---
`NOTE`:

You want to lemmatize before stemming because lemmatization requires real dictionary words to work properly, whereas stemming cuts off the ends of words without checking a dictionary.

If you reverse the order, the stemmer will destroy the structure of the words, leaving the lemmatizer completely useless.

### The Breakdown
- Lemmatizer (The Smart Tool): It looks up words in a dictionary (like WordNet) to find their base linguistic form. For example, it transforms "studies" to "study".

- Stemmer (The Brutal Tool): It applies harsh algorithmic rules to chop off suffixes. For example, it chops "studies" or "study" down to "studi"

### What happens if you do Stemming first?

1. The raw word is "studies".
2. Stemmer runs first: It chops "studies" down to the root "studi".
3. Lemmatizer runs second: It looks up "studi" in its linguistic dictionary. Because "studi" is not a real English word, the lemmatizer fails to recognize it and does nothing.

### What happens if you do Lemmatization first (correct way)?

1. The raw word is "studies".
2. Lemmatizer runs first: It looks up "studies" in the dictionary and neatly normalizes it to the real word "study".
3. Stemmer runs second: It takes the clean word "study" and cuts it down to its core machine-learning root "studi".


By placing lemmatization first, you give your text processing pipeline the highest possible accuracy before the stemmer reduces everything to rough machine-readable roots.
