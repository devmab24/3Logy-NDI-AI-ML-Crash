```text
Build a simple, professional Streamlit UI for our project:

SMART INCIDENT REPORT ANALYZER (SIRA)

SIRA is a machine-learning application that analyzes workplace incident reports and predicts the type/category of incident using a trained classification model.

The application is currently a Python ML project with this general architecture:

smart_incident_report_analyzer/
│
├── data/
├── notebooks/
├── src/
│   ├── data_loader.py
│   ├── preprocessing.py
│   ├── data_splitter.py
│   ├── feature_engineering.py
│   ├── model.py
│   └── models/
│       ├── logistic_regression.py
│       ├── naive_bayes.py
│       ├── svm_classifier.py
│       ├── decision_tree.py
│       └── random_forest.py
│
├── models/
├── train.py
├── benchmark.py
├── evaluate.py
├── main.py
└── requirements.txt

The application should NOT be a complex enterprise dashboard. It should be a simple MVP that demonstrates how a machine-learning model can be exposed through a web interface.

### UI REQUIREMENTS

Create a single-page Streamlit application.

The interface should contain:

1. HEADER

Display:

"Smart Incident Report Analyzer"

Subtitle:

"AI-powered workplace incident classification"

Add a short description:

"SIRA analyzes incident reports and predicts the most likely incident category using machine learning."

2. SIDEBAR

Create a simple sidebar containing:

- SIRA logo/title
- "Model Information"
- Model name
- Model status
- Application version

For now, use placeholder values where necessary.

Example:

Model:
Logistic Regression

Status:
Ready

Version:
v1.0.0

Do not create complicated navigation.

3. INCIDENT REPORT INPUT

The main section should contain:

"Enter Incident Report"

Use a large Streamlit text_area where the user can enter an incident report.

Example placeholder:

"Example: An employee slipped on a wet floor while walking through the warehouse..."

Add a "Analyze Incident" button.

4. PREDICTION RESULT

When the user clicks "Analyze Incident":

Display a clear result section:

"Prediction"

Then display:

Predicted Incident Type:
[incident type]

Confidence:
[confidence percentage]

For now, if the trained model does not support probability prediction, handle this gracefully instead of breaking the application.

Use an appropriate message such as:

"Confidence score is not available for this model."

5. INCIDENT SUMMARY

Create a simple section called:

"Incident Analysis"

For the MVP, display:

- Predicted incident type
- Confidence (if available)
- Report length
- Number of words

Do NOT implement generative AI yet.

6. RECENT ANALYSES

Create a small section showing the reports analyzed during the current Streamlit session.

Use st.session_state to maintain the history.

Display a simple table containing:

- Incident Report
- Prediction
- Confidence

The history should disappear when the Streamlit session is restarted. No database is required.

7. EMPTY STATE

When no report has been entered, display a simple message such as:

"Enter an incident report above to begin analysis."

8. ERROR HANDLING

The application should gracefully handle:

- Empty incident reports
- Missing model files
- Prediction errors
- Preprocessing errors

Use Streamlit warnings/errors rather than allowing the application to crash.

9. DESIGN

Keep the design clean and minimal.

Use:

- Streamlit native components
- st.title()
- st.header()
- st.subheader()
- st.text_area()
- st.button()
- st.metric()
- st.info()
- st.success()
- st.warning()
- st.error()
- st.dataframe()

Avoid:

- Complex animations
- Excessive cards
- Complex navigation
- Authentication
- Databases
- APIs
- Chatbots
- Generative AI
- AWS integration
- Complex dashboards

The purpose is to build the first functional UI around the existing ML pipeline.

10. ML INTEGRATION

The UI should be designed so that the ML pipeline can be connected cleanly.

Do NOT put all machine-learning logic directly inside app.py.

Create a simple separation:

app.py
    ↓
UI layer

src/
    ↓
Preprocessing
Feature engineering
Model loading
Prediction

The Streamlit application should eventually perform:

User enters report
        ↓
Text preprocessing
        ↓
TF-IDF transformation
        ↓
Trained model
        ↓
Prediction
        ↓
Display result

Use st.cache_resource() when loading the trained model/vectorizer so that they are not reloaded every time the user interacts with the page.

11. PROJECT STRUCTURE

Recommend a clean structure such as:

smart_incident_report_analyzer/
│
├── app.py
├── train.py
├── benchmark.py
├── evaluate.py
├── requirements.txt
│
├── models/
│   ├── classifier.pkl
│   └── vectorizer.pkl
│
└── src/
    ├── __init__.py
    ├── preprocessing.py
    ├── feature_engineering.py
    └── predictor.py

Create predictor.py as a small abstraction responsible for:

- Loading the trained model
- Loading the TF-IDF vectorizer
- Preprocessing incoming text
- Transforming the text
- Making predictions
- Returning the prediction result

The Streamlit app should call the predictor rather than directly manipulating the sklearn model.

12. CODE QUALITY

Write clean, beginner-friendly Python.

Use functions where appropriate.

Add comments explaining important sections.

Do not over-engineer the application.

The students should be able to understand:

- How Streamlit works
- How widgets collect user input
- How button events work
- How session state works
- How an ML model is connected to a UI
- How preprocessing and prediction happen
- How prediction results are displayed

13. IMPORTANT

The current SIRA model was trained using:

TF-IDF
+
Incident classification model

The benchmark currently compares:

- Logistic Regression
- Naive Bayes
- Linear SVM
- Decision Tree
- Random Forest

The UI should therefore NOT hardcode assumptions that the final production model must always be Logistic Regression.

Design the predictor so that the trained model can be replaced later.

The current benchmark results are experimental and should not be presented as production performance.

14. FINAL OUTPUT

Provide:

1. Recommended project structure
2. Complete app.py
3. Complete src/predictor.py
4. Any required changes to preprocessing.py or feature_engineering.py
5. requirements.txt additions
6. Instructions for saving/loading the trained model and TF-IDF vectorizer
7. Command to run the application:

streamlit run app.py

Explain briefly how data flows through the application.

Keep the implementation simple enough for students learning ML engineering and Streamlit.
```
