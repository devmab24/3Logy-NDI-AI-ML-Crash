## Challenge

You are challenged to build an end-to-end preprocessing pipeline:

1. Load the `incident_reports.csv` dataset using `DataLoader`.
2. Use `DataExplorer` to inspect the data and report any quality issues.
3. Clean the dataset by handling missing values, duplicates, whitespace, and capitalization.
4. Create a `TextPreprocessor` class that:

   * Converts text to lowercase.
   * Removes punctuation.
   * Removes numbers (if appropriate).
   * Removes extra spaces.
   * Tokenizes the text.
   * Removes stop words.
   * Applies lemmatization.
5. Create a `FeatureEngineer` class that:

   * Generates Bag of Words features using `CountVectorizer`.
   * Generates TF-IDF features using `TfidfVectorizer`.
6. Compare the feature matrices and answer:

   * How many unique words are in the vocabulary?
   * Which representation is more suitable for text classification?
   * Why must the same preprocessing pipeline be used during both training and prediction?
