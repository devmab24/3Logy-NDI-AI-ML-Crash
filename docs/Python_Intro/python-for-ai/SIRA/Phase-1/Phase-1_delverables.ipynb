### Phase 1 deliverable

## Project Title

**Smart Incident Report Analyzer (SIRA)**

### Goal

Build the **foundation** of a production-ready AI/ML project.

Students should understand:

* How professional ML projects are organized.
* Why each file and folder exists.
* How modules communicate.
* How OOP fits into an ML project.
* How to run a multi-file Python application.

---

### Final deliverable

Students should submit a working project with the following structure:

```text
smart_incident_report_analyzer/

│
├── data/
│   ├── raw/
│   ├── processed/
│   └── external/
│
├── notebooks/
│   └── 01_python_to_ml_engineering.ipynb
│
├── src/
│   ├── __init__.py
│   ├── incident.py
│   ├── data_loader.py
│   └── utils.py
│
├── models/
│
├── main.py
├── requirements.txt
├── README.md
└── .gitignore
```

---

### What Each File Should Do

### `incident.py`

Contains the main class.

Example:

```python
class Incident:

    def __init__(self, incident_id, report):

        self.incident_id = incident_id
        self.report = report

    def display(self):

        print(f"{self.incident_id}: {self.report}")
```

Students demonstrate:

* Classes
* Constructors
* Instance attributes
* Methods

---

### `data_loader.py`

Contains only functions related to loading data.

Example:

```python
def load_sample_reports():
    ...
```

Later this file will read CSV files.

For now it can return sample incident reports.

---

### `utils.py`

Contains helper functions.

Example:

```python
def print_banner():

    print("=" * 50)

    print("Smart Incident Report Analyzer")

    print("=" * 50)
```

Students learn:

* Reusable functions
* Module imports

---

### `main.py`

The application entry point.

Example workflow:

```text
Start Application

↓

Display Banner

↓

Load Sample Reports

↓

Create Incident Objects

↓

Display Reports

↓

Application Ends
```

Students now see how multiple files work together.
