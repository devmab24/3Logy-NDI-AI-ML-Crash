### 1. **SageMaker AI**

The `Amazon SageMaker AI` (formerly known simply as Amazon SageMaker) is a tool used to train, build and deploy machine learning models. Amazon SageMaker AI is a fully managed cloud service by Amazon Web Services (AWS) that enables developers and data scientists to build, train, and deploy machine learning (ML) and artificial intelligence (AI) models at scale. 
The `AWS SageMaker` is an alternative but it is specifically used for data analytics and AI.

Steps to create a **Jupyter Notebook**:

* From the Search Menu, type `SageMaker AI` and click on the search button
* From the left side menu lists:
    * Find `Applications and IDEs` list.
    * Select **Notebooks**.
    * Select the `Notebooks Instances` tab. 
* On the Console/UI Click  on `Create notebook instance`

> Note: When you click "Create notebook instance" in Amazon SageMaker, AWS handles a multi-step orchestration process behind the scenes. It sets up the physical hardware, configures the networking, loads the machine learning software, and establishes security bridges.

**Here is the exact step-by-step breakdown of what happens under the hood:**

* **Spins up a server**: AWS launches a dedicated virtual machine (EC2 instance) based on the size you chose.
* **Attaches a hard drive**: It links a persistent storage volume (EBS) to save your files when the machine is turned off.
* **Loads the software**: It runs a container pre-packaged with Linux, JupyterLab, and AI libraries like PyTorch and TensorFlow.
* **Secures the network**: It applies your IAM permissions so you can securely access Amazon S3 data without typing passwords.
* **Creates a secure link**: It generates an encrypted web URL so you can log into your workspace from your browser.

> A SageMaker **notebook instance** is a fully managed cloud computer (virtual machine) that comes pre-installed with Jupyter Notebook applications and machine learning libraries. It gives data scientists, and ML Engineers a ready-to-use environment to write code, explore data, and build models without setting up any local software.

*Let's continue;*

* Fill the information required to complete the instance creation.
    * Write the `Notebook instance name` e.g `sira development course`.
    * Notebook instance type e.g `ml.t3.medium`. These are shape and size which will dictate how much CPU, how much Memory and how much storage behind the scene it should allocate.
    * Leave the default `Platform identifier`.
    * Root access is optional.
        * You can `enable` or `disable` Access. (Optional)
            * If you wish to create a role then click on `Create role using the role creation wizard` and complete setup.
    * Click `Create notebook instance`.
    * Finally, Open the notebook instance when status is `InService` and open a template notebook to get started.

> A **Jupyter Notebooks**  is the primary interactive document where you write and run your data science, machine learning, and AI code.
Instead of writing a standard Python script that runs all at once, a notebook allows you to divide your work into independent "cells". You can execute one cell to pull data, another to train a model, and a third to instantly print a chart-all within the same file.

> **JupyterLab** is a web-based IDE for your notebooks, code, and data. Create a dedicated environment with your preferred configuration and storage to start developing ML projects.

> JupyterLab serves as the primary workspace where you spend your time writing Python and machine learning code.

> Amazon **SageMaker Studio** is a fully managed, web-based Integrated Development Environment (IDE) designed specifically for machine learning and artificial intelligence workflows.

Instead of jumping between different tools to clean data, write code, track experiments, and deploy models, SageMaker Studio unifies the entire machine learning lifecycle into a single graphical user interface
