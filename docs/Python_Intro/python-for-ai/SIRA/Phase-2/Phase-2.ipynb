### New `module`

You may think where should

```python
pd.read_csv(...)
```

go?

Remember not `main.py`, it belongs inside

```text
src/

data_loader.py
```

Why?

Because, loading data is one responsibility.

---

Update:

data_loader.py

```python
import pandas as pd
from pathlib import Path


class DataLoader:
    """
    Responsible for loading datasets.

    This class has only one responsibility:
    Reading datasets from storage.
    """

    def __init__(self):

        self.data_path = Path("data/raw/incident_reports.csv")

    def load_data(self):

        df = pd.read_csv(self.data_path)

        return df
```

### Understanding the `data`

We won't jump into cleaning yet, instead lets explore;

```python
- df.head() `First impression`
- df.tail() `Last records`
- df.shape `Rows&Columns`
- df.columns `Column names`
- df.info() `Data types, missing values. Memory usage`
- df.describe(include="all") `Statistics`
- df.sample(5) - `Random inspection`

```

Why all these?

Always know that doing this is very important. Engineers should always inspect before cleaning. Professionals do this constantly.

### Create another module

Now you understand loading. Next is to create

```text
src/

eda.py
```

Why?

Should `DataLoader` also perform analysis? No. Again `Single responsibility`.

---

Example

```python
class DataExplorer:

    def basic_summary(self, df):

        print(df.info())

        print(df.describe(include="all"))
```

Notice

One responsibility.

---

### Update main.py

```python
from src.data_loader import DataLoader
from src.eda import DataExplorer


def main():

    loader = DataLoader()

    explorer = DataExplorer()

    df = loader.load_data()

    explorer.basic_summary(df)


if __name__ == "__main__":
    main()
```

### What students have learned in `Phase 2`

| Concept                         | Why it Matters                                                                                                |
| ------------------------------- | ------------------------------------------------------------------------------------------------------------- |
| Data acquisition                | Every ML project begins with data collection.                                                                 |
| Project data organization       | Separating raw, processed, and external data improves reproducibility and prevents accidental data loss.      |
| `pathlib.Path`                  | Creates portable code that works across operating systems.                                                    |
| `DataLoader` class              | Encapsulates all data access logic, making it easy to swap data sources later (CSV, database, cloud storage). |
| Pandas data loading             | The standard way to ingest tabular datasets in Python.                                                        |
| Exploratory Data Analysis (EDA) | Helps understand the structure, quality, and characteristics of the data before making changes.               |
| Separation of concerns          | `DataLoader` loads data; `DataExplorer` analyzes data. Each module has a single, clear responsibility.        |
